```python
import pandas as pd
import matplotlib.pyplot as plt

CampsRaw = 'https://raw.githubusercontent.com/cta48/GeoData-Soil-Moisture-Sensors/refs/heads/main/CAMPS.csv'
CampsData = pd.read_csv(CampsRaw, skiprows=4)
CampsData["timestamp"] = pd.to_datetime(CampsData["timestamp"], errors="coerce")
CampsData["volumetric_water_content[%]"] = pd.to_numeric(CampsData["volumetric_water_content[%]"], errors="coerce")
CampsData[" vwc_raw[V]"] = pd.to_numeric(CampsData[" vwc_raw[V]"], errors="coerce")
CampsData["battery[V]"] = pd.to_numeric(CampsData["battery[V]"], errors="coerce")


CenseRaw = 'https://raw.githubusercontent.com/cta48/GeoData-Soil-Moisture-Sensors/refs/heads/main/CENSE.csv'
CenseData = pd.read_csv(CenseRaw, skiprows=4)
CenseData["timestamp"] = pd.to_datetime(CenseData["timestamp"], errors="coerce")
CenseData["volumetric_water_content[%]"] = pd.to_numeric(CenseData["volumetric_water_content[%]"], errors="coerce")
CenseData[" vwc_raw[V]"] = pd.to_numeric(CenseData[" vwc_raw[V]"], errors="coerce")
CenseData["battery[V]"] = pd.to_numeric(CenseData["battery[V]"], errors="coerce")


GlitzRaw = 'https://raw.githubusercontent.com/cta48/GeoData-Soil-Moisture-Sensors/refs/heads/main/GLITZ.csv'
GlitzData = pd.read_csv(GlitzRaw, skiprows=4)
GlitzData["timestamp"] = pd.to_datetime(GlitzData["timestamp"], errors="coerce")
GlitzData["volumetric_water_content[%]"] = pd.to_numeric(GlitzData["volumetric_water_content[%]"], errors="coerce")
GlitzData[" vwc_raw[V]"] = pd.to_numeric(GlitzData[" vwc_raw[V]"], errors="coerce")
GlitzData["battery[V]"] = pd.to_numeric(GlitzData["battery[V]"], errors="coerce")

GraspRaw = 'https://raw.githubusercontent.com/cta48/GeoData-Soil-Moisture-Sensors/refs/heads/main/GRASP.csv'
GraspData = pd.read_csv(GraspRaw, skiprows=4)
GraspData["timestamp"] = pd.to_datetime(GraspData["timestamp"], errors="coerce")
GraspData["volumetric_water_content[%]"] = pd.to_numeric(GraspData["volumetric_water_content[%]"], errors="coerce")
GraspData[" vwc_raw[V]"] = pd.to_numeric(GraspData[" vwc_raw[V]"], errors="coerce")
GraspData["battery[V]"] = pd.to_numeric(GraspData["battery[V]"], errors="coerce")

NinjaRaw = 'https://raw.githubusercontent.com/cta48/GeoData-Soil-Moisture-Sensors/refs/heads/main/NINJA.csv'
NinjaData = pd.read_csv(NinjaRaw, skiprows=4)
NinjaData["timestamp"] = pd.to_datetime(NinjaData["timestamp"], errors="coerce")
NinjaData["volumetric_water_content[%]"] = pd.to_numeric(NinjaData["volumetric_water_content[%]"], errors="coerce")
NinjaData[" vwc_raw[V]"] = pd.to_numeric(NinjaData[" vwc_raw[V]"], errors="coerce")
NinjaData["battery[V]"] = pd.to_numeric(NinjaData["battery[V]"], errors="coerce")


# Putting them ina dict so its easier to loop
datasets = {
    "CAMPS": CampsData,
    "GLITZ": GlitzData,
    "CENSE": CenseData,
    "GRASP": GraspData,
    "NINJA": NinjaData
    }

print("----------Camps Data----------")
print(CampsData.head())
print("----------Cense Data----------")
print(CenseData.head())
print("----------Glitz Data----------")
print(GlitzData.head())
print("----------Grasp Data----------")
print(GraspData.head())
print("----------Ninja Data----------")
print(NinjaData.head())
```

```python
# Plot Data against Time

for name, df in datasets.items():
    plt.figure(figsize=(12, 5))
    plt.plot(df["timestamp"], df["volumetric_water_content[%]"])
    plt.xlabel("Time")
    plt.ylabel("VWC (%)")
    plt.title(f"{name}: VWC vs Time")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```
